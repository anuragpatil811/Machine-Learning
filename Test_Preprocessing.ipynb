```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
sen = ['He is Abhishek', 'Abhishek is second in the class', 'Ritesh is third in the class', 'Is he Abhishek?' ]
```

```python
#Count Vectorizer
cv = CountVectorizer()
```

```python
cv.fit(sen)
```

```python
cv.vocabulary_
```

```python
cv.get_feature_names_out()
```

```python
x = cv.fit_transform(sen)
```

```python
x.toarray()
```

```python
pd.DataFrame(data=x.toarray(), columns = cv.get_feature_names_out())
```

```python
#TF-IDF Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
tf = TfidfVectorizer()
```

```python
tf.fit(sen)
```

```python
y = tf.fit_transform(sen)
```

```python
y.toarray()
```

```python
tf.get_feature_names_out()
```

```python
pd.DataFrame(data = y.toarray(), columns=tf.get_feature_names_out())
```
