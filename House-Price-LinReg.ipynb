```python
from sklearn.preprocessing import LabelEncoder
enc = LabelEncoder()
```

```python
df["Town"] = enc.fit_transform(df["Town"])
```

```python
df["Town"].head()
```

```python
x = df.drop("Price", axis=1)
y = df["Price"]
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
y_predict = model.predict(x)
y_predict
```

```python
r2_score(y, y_predict)
```
