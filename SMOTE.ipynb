```python
import pandas as pd 
```

```python
data = {
    "Age": [22, 25, 47, 52, 46, 56, 23, 40, 60, 48, 90, 39, 49],
    "Salary": [20000, 25000, 50000, 52000, 48000, 60000, 22000, 45000, 65000, 49000, 45000, 59000, 39000],
    "Purchased": [0, 0, 1, 1, 1, 1, 0, 1, 1, 1, 0, 0, 0]
}
df= pd.DataFrame(data)
```

```python
df 
```

```python
df["Purchased"].value_counts()
```

```python
df.duplicated().sum()
```

```python
x = df.drop("Purchased", axis=1)
y = df["Purchased"]
```

```python
from imblearn.over_sampling import RandomOverSampler 
```

```python
rv = RandomOverSampler()
```

```python
rvx, rvy = rv.fit_resample(x, y)
```

```python
rvy.value_counts()
```

```python
rvx.duplicated().sum()
```

```python
#Undersampling 
x = df.drop("Purchased", axis=1)
y = df["Purchased"]
from imblearn.under_sampling import RandomUnderSampler
```

```python
rs = RandomUnderSampler()
```

```python
rsx, rsy = rs.fit_resample(x, y)
```

```python
rsy.value_counts()
```

```python
rsx.duplicated().sum()
```

```python
#SMOTE
from imblearn.over_sampling import SMOTE 
```

```python
sm = SMOTE()
```

```python
smx, smy = sm.fit_resample(x, y)
```

```python
smy.value_counts()
```

```python
smx.duplicated().sum()
```
